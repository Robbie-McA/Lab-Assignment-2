## Part 3 - Creating Static Chloropleth Maps with seperate Classification Methods



```python
# Setting the number of classes used for each classification
class_count = 5

# Setting the Quantiles classification, splitting data into classes each with an equal number of features
classifier_q = mc.Quantiles(SIMD_EdinburghCity['IncNumDep'], k=class_count)

# Setting the Equal Interval Mean, splitting data into classes based on their values
classifier_ei = mc.EqualInterval(SIMD_EdinburghCity['IncNumDep'], k=class_count)
```

```python
# Creating Histograms of this Income Deprivation for both quantile and equal interval classification
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Creating the histogram for the quantile classification
sns.histplot(data=SIMD_EdinburghCity, x="IncNumDep", ax=axes[0], kde=True, bins=20)
axes[0].axvline(classifier_q.bins[0], color='red', linestyle='dashed', linewidth=2, label='Quantiles')
for bin_value in classifier_q.bins:
    axes[0].axvline(bin_value, color='red', linestyle='dashed', linewidth=2)
axes[0].set_title("IncNumDep Histogram with Quantile Breaks")
axes[0].legend()

# Creating the histogram for the equal interval classification
sns.histplot(data=SIMD_EdinburghCity, x="IncNumDep", ax=axes[1], kde=True, bins=20)
axes[1].axvline(classifier_ei.bins[0], color='blue', linestyle='dashed', linewidth=2, label='Equal Interval Breaks')
for bin_value in classifier_ei.bins:
    axes[1].axvline(bin_value, color='blue', linestyle='dashed', linewidth=2)
axes[1].set_title("IncNumDep Histogram with Equal Interval Breaks")
axes[1].legend()

# Presenting the histograms in the same output
plt.tight_layout()
plt.show()
```

```python
# Setting figure size to include 2 figures in the one output as subplots
fig, axs = plt.subplots(1, 2, figsize=(18, 8))

# Creating chloropleth Map for the Income deprivation in Edinburgh using the quantile classification
SIMD_EdinburghCity.plot(column='IncNumDep', ax=axs[0],
         legend=True, cmap='viridis',
         scheme='UserDefined',
         classification_kwds={'bins': classifier_q.bins}
        )

axs[0].set_title("Choropleth Map for Edinburgh Cities Income Deprivation with Quantile Breaks")

# Creating chloropleth Map for the Income deprivation in Edinburgh using the equal interval classification
SIMD_EdinburghCity.plot(column='IncNumDep', ax=axs[1],
         legend=True, cmap='viridis',
         scheme='UserDefined',
         classification_kwds={'bins': classifier_ei.bins})

axs[1].set_title("Choropleth Map for Edinburgh City's Income Deprivation with Equal Interval Breaks")

plt.tight_layout()
plt.show()
```

### Differences between Classification Technique Outputs

The Quantile classification has far more representation of the upper values of income deprivation due to that final class spanning the largest range of values by far compared to the rest, as each class has an equal amount of counts. Meanwhile, the Equal Interval classification, being balanced with each equal weighting given to each range of values leaned far more towards the lower values of income deprivation with the highest counts.

```python
## Loading the API and setting the bounding box to Dundee
OHSOME_URL = "https://api.heigit.org/ohsome-api/v2-rc/stats/features/"
BBOX = [-3.033772,56.484582,-2.936611,56.516037]

# Asking Ohsome for the data within our North Dundee Bounding box, for data within our filter between january 2008 and january 2026
def ohsome(measure, osm_filter):
    body = {"aoi": BBOX,
            "filter": osm_filter,
            "time": {"start": "2008-01-01", "end": "2026-01-01", "interval": "P1Y"}}
    response = requests.post(OHSOME_URL + measure + ".json", json=body,
                             headers={"Authorization": API_KEY}, timeout=300)
    response.raise_for_status()           
    result = response.json()["result"]
    years = [int(t[:4]) for t in result["timestamp"]]
    return pd.Series(result["value"], index=years)

# Defining our buildings and streets data from Ohsome
buildings = ohsome("count", "building=* and geometry:polygon")
streets = ohsome("length", "highway in (residential, living_street) and type:way") / 1000   # metres -> km

# Creating our dataframe
growth = pd.DataFrame({"Buildings (count)": buildings, "Residential streets (km)": streets})
growth
```

```python
# Creating histograms of our building counts and street counts data
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
growth["Buildings (count)"].plot(ax=axes[0], marker="o", color="#00539B", title="Buildings on the map")
growth["Residential streets (km)"].plot(ax=axes[1], marker="o", color="#785596", title="Residential streets on the map (km)")
for ax in axes:
    ax.set_ylim(bottom=0)
    ax.set_xlabel("Year")
    ax.xaxis.set_major_locator(plt.MaxNLocator(integer=True))   # whole years on the axis
plt.tight_layout()
plt.show()

# Calculating the growth in mapped buildings and streets
change = (growth.loc[2026] / growth.loc[2012] - 1) * 100
print("Change from 2012 to 2026 (%):")
print(change.round(0))
```

The patterns repeat between the Edinburgh example and Dundee above, however the trend repeats for residential streets at the same time as Edinburgh, beginning at 2008, with buildings having their sudden rise more than a decade after the Edinburgh graph.

### Part 1

```python
# Setting the walking speed variable at 80 metres per minute
WALK_SPEED = 80                  

# Loading in our dundee walk dataset, setting the crs and creating nodes and edges
G = ox.load_graphml("data/dundee_walk.graphml")
G = ox.project_graph(G, to_crs="EPSG:27700")           
nodes, edges = ox.graph_to_gdfs(G)

# Loading in the services geopackage
dundee_services = gpd.read_file("data/dundee_services.gpkg").to_crs("EPSG:27700")

# Slicing the dataset to only include schools and printing the number of street junctions and schools
schools = dundee_services[dundee_services["service"] == "School"].copy()
print(len(nodes), "street junctions and", len(schools), "schools")
```

```python
# Adding a nodes column to the schools data to allow for mapping
schools["node"] = ox.distance.nearest_nodes(G, schools.geometry.x, schools.geometry.y)
schools[["name", "node"]].head()
```

```python
# Finding nearest schools to each junction
nearest = nx.multi_source_dijkstra_path_length(G, set(schools["node"]), weight="length")
nodes["nearest_min"] = pd.Series(nearest) / WALK_SPEED

# Plotting Dundee nodes with nearest schools mapped
ax = nodes.plot(column="nearest_min", cmap="Oranges", vmax=20, markersize=2, legend=True, figsize=(8, 8),
                legend_kwds={"label": "Minutes to nearest school"})
schools.plot(ax=ax, color="black", markersize=12)
ax.set_title("Measure 1: nearest school (black dots = schools)")
ax.set_axis_off()
plt.show()
```

```python
# Setting the beta parameter for gravity model, based on 10-minute half-life (ln(2)/10)
BETA = 0.069 

# Setting values and weighting for each node based on their distance to nearest schools
cumulative = {node: 0 for node in G.nodes}     
gravity = {node: 0.0 for node in G.nodes}

for school_node in schools["node"]:
    distances = nx.single_source_dijkstra_path_length(G, school_node, cutoff=30 * WALK_SPEED, weight="length")
    for node, metres in distances.items():
        minutes = metres / WALK_SPEED
        if minutes <= 10:
            cumulative[node] += 1
        gravity[node] += math.exp(-BETA * minutes)

nodes["cumulative"] = pd.Series(cumulative)
nodes["gravity"] = pd.Series(gravity)

nodes[["nearest_min", "cumulative", "gravity"]].describe().round(2)
```

```python
# Plotting the two measures side by side
fig, axes = plt.subplots(1, 2, figsize=(14, 6.5))
nodes.plot(ax=axes[0], column="cumulative", cmap="Blues", markersize=2, legend=True)
axes[0].set_title("Measure 2: schools within a 10-minute walk")
nodes.plot(ax=axes[1], column="gravity", cmap="Purples", markersize=2, legend=True)
axes[1].set_title("Measure 3: gravity score")
for ax in axes:
    ax.set_axis_off()
plt.show()
```

```python
# Setting nodes and edges for dataset
walk_nodes, walk_edges = ox.graph_to_gdfs(G_walk)

# importing the residential dataset
residential = gpd.read_file("data/perth_residential.gpkg").to_crs("EPSG:27700")
residential_area = unary_union(list(residential.geometry))

# Setting the centre point of the perth area using the coordinates submitted earlier
centre = gpd.GeoSeries.from_xy([PERTH[1]], [PERTH[0]], crs="EPSG:4326").to_crs("EPSG:27700")
neighbourhood = centre.buffer(AREA).iloc[0]

# Setting homes as nodes within the residential dataset
homes = walk_nodes[walk_nodes.within(residential_area) & walk_nodes.within(neighbourhood)].copy()
print(len(homes), "residential junctions in the neighbourhood")
```

```python
# Importing the services dataset
services = gpd.read_file("data/perth_services.gpkg").to_crs("EPSG:27700")
homes_on_bike = ox.distance.nearest_nodes(G_bike, homes.geometry.x, homes.geometry.y) 

# Setting services to locations and returning number of each service
for name in SERVICES:
    points = services[services["service"] == name].geometry

    walk_targets = ox.distance.nearest_nodes(G_walk, points.x, points.y)
    bike_targets = ox.distance.nearest_nodes(G_bike, points.x, points.y)
    walk_dist = nx.multi_source_dijkstra_path_length(G_walk, set(walk_targets), weight="length")
    bike_dist = nx.multi_source_dijkstra_path_length(G_bike, set(bike_targets), weight="length")

    homes[name + " walk (m)"] = [walk_dist.get(n) for n in homes.index]
    homes[name + " bike (m)"] = [bike_dist.get(n) for n in homes_on_bike]
    print(f"{name:<17} {len(points):>5} found")
```

```python
# Creating the dataset to show percentage of services within 15 mins walking and 20 mins biking
results = pd.DataFrame(index=list(SERVICES))
results["% homes within 15-min walk"] = [(homes[s + " walk (m)"] <= WALK_LIMIT).mean() * 100 for s in SERVICES]
results["% homes within 20-min cycle"] = [(homes[s + " bike (m)"] <= BIKE_LIMIT).mean() * 100 for s in SERVICES]
results.round(0)
```

```python
# Calculating services within the 15 min walking distance from homes
homes["services_walk"] = sum((homes[s + " walk (m)"] <= WALK_LIMIT).astype(int) for s in SERVICES)

# Plotting the homes with access to all 5 services
ax = homes.plot(column="services_walk", cmap="RdYlGn", vmin=0, vmax=5, markersize=6, legend=True,
                figsize=(8, 8), legend_kwds={"label": "Essential services within a 15-minute walk (of 5)"})
gpd.GeoSeries([neighbourhood], crs="EPSG:27700").boundary.plot(ax=ax, color="grey")
ax.set_title("Perth: is it a 15-minute neighbourhood?")
ax.set_axis_off()
plt.show()

print(f"{(homes['services_walk'] == 5).mean():.0%} of residential junctions reach all 5 services on foot")
```

Only 30% of Homes are within the distance when changed to 800 metres

```python
# Setting the catchment distance
Catchment = 1200
```

```python
# Snapping parks and buildings to the network
parks["node"] = ox.distance.nearest_nodes(G_city, parks.representative_point().x, parks.representative_point().y)
buildings["node"] = ox.distance.nearest_nodes(G_city, buildings.representative_point().x, buildings.representative_point().y)

# Creating the demand from the number of buildings at each junction
demand = buildings["node"].value_counts()
demand.head()
```

```python
reach = {}          # which junctions can walk to each park
ratio = {}          # park area divided by the buildings that can reach it

# Setting park area per building
for park_id, park in parks.iterrows():
    nearby = nx.single_source_dijkstra_path_length(G_city, park["node"], cutoff=Catchment, weight="length")
    buildings_nearby = sum(demand.get(node, 0) for node in nearby)
    reach[park_id] = nearby
    ratio[park_id] = park["area_m2"] / buildings_nearby if buildings_nearby > 0 else 0

parks["m2_per_building"] = pd.Series(ratio)
parks.sort_values("m2_per_building", ascending=False)[["name", "area_m2", "m2_per_building"]].head(10).round(0)
```

```python
city_nodes, city_edges = ox.graph_to_gdfs(G_city)
park_access = {node: 0.0 for node in G_city.nodes}

# Adding up available park space for each home
for park_id, nearby in reach.items():
    for node in nearby:
        park_access[node] += ratio[park_id]

city_nodes["park_access"] = pd.Series(park_access)

homes_city = city_nodes[city_nodes.index.isin(demand.index)]
homes_city["park_access"].describe().round(0)
```

```python
# Plotting Park area per building within a 15 minute walk
fig, ax = plt.subplots(figsize=(9, 9))
city_edges.plot(ax=ax, color="lightgrey", linewidth=0.4)
parks.plot(ax=ax, color="#a8d5a2")
homes_city.sort_values("park_access").plot(ax=ax, column="park_access", cmap="Greens", scheme="quantiles", k=5,
                                           markersize=8, legend=True,
                                           legend_kwds={"title": "Park m² per building (2SFCA)", "loc": "lower left"})
ax.set_title(f"{CITY}: park space within a 15-minute walk, shared with neighbours")
ax.set_axis_off()
plt.show()

# Printing the percentage of homes with no park within 15 minutes of walking
print(f"{(homes_city['park_access'] == 0).mean():.0%} of homes have no park within a 15-minute walk")
```

```python
# Calculating the percentages and saving them in a result dataframe based upon the values being analysed and their total columns
def calculate_percentages(dataframe, total_columns, value_columns):

    result_df = pd.DataFrame()

    for total_col, value_col in zip(total_columns, value_columns):
        percentage_col_name = f"{value_col}_percentage"

        if total_col not in dataframe.columns:
            raise ValueError(f"Total column '{total_col}' not found in the DataFrame.")
        # The data have several empty values, so you need to process them, forcing them to be a numeric or NaN value
        dataframe[value_col] = pd.to_numeric(dataframe[value_col], errors='coerce')
        dataframe[total_col] = pd.to_numeric(dataframe[total_col], errors='coerce')
        
        result_df[percentage_col_name] = (dataframe[value_col] / dataframe[total_col]) * 100

    return result_df

# List of the total columns
total_cols = ['All people', 'All people', 'All people',
              'All people.1', 'All people.1', 'All people.1',
              'All people aged 16 to 74', 'All people aged 16 to 74', 'All people aged 16 to 74',
              'All people aged 16 to 74 in employment', 'All people aged 16 to 74 in employment', 'All people aged 16 to 74 in employment', 'All people aged 16 to 74 in employment',
              'All people.2', 'All people.2',
              'All households', 'All households', 'All households',
              'All people.3', 'All people.3', 'All people.3',
              'All people.4', 'All people.4', 'All people.4',
              'All people aged 16 and over', 'All people aged 16 and over',
              'All household reference persons aged 16 to 64', 'All household reference persons aged 16 to 64']
# List of the value columns
value_cols = ['20 to 24',
              '30 to 44',
              '75 to 84',
              'White: Scottish',
              'Asian, Asian Scottish or Asian British',
              'African: African, African Scottish or African British',
              'Economically active: Employee: Full-time',
              'Economically active: Employee: Part-time',
              'Unemployed people aged 16 to 74: Long-term unemployed',
              'F. Construction',
              'I. Accommodation and food service activities',
              'K. Financial and insurance activities',
              'P. Education',
              'Males',
              'Females',
              'Household is not deprived in any dimension',
              'Household is deprived in 2 dimensions',
              'Household is deprived in 4 dimensions',
              'Good health',
              'Bad health',
              'Very bad health',
              'Day-to-day activities not limited',
              'Day-to-day activities limited a little',
              'Day-to-day activities limited a lot',
              'All people aged 16 and over: No qualifications',
              'All people aged 16 and over: Level 4 and above',
              'DE Semi-skilled and unskilled manual workers; on state benefit, unemployed, lowest grade workers',
              'AB Higher and intermediate managerial/administrative/professional']

result_dataframe = calculate_percentages(merged_data, total_cols, value_cols)
```

### Part 8 - Defining optimum Cluster (K) value, using elbow technique

```python
# Creating the sum of squared distances dataset
Sum_of_squared_distances = []

# Setting a cluster range of 15
K_range = range(1,15)

# Returning the sum of squared distances from the centroid
for k in K_range:
 km = KMeans(n_clusters=k)
 km = km.fit(z_score_df)
 Sum_of_squared_distances.append(km.inertia_)

# Plotting the elbow graph
plt.plot(K_range, Sum_of_squared_distances, 'bx-')
plt.xlabel('k')
plt.ylabel('Sum_of_squared_distances')
plt.title('Elbow Method For Optimal k')
plt.show()
```

```python
# Graphing the between cluster sum of squares
import numpy as np
def elbow(dataframe, n):
    kMeansVar = [KMeans(n_clusters=k).fit(dataframe.values) for k in range(1, n)] #making use of list comprehensions.
    centroids = [X.cluster_centers_ for X in kMeansVar]
    k_euclid = [cdist(dataframe.values, cent) for cent in centroids]
    dist = [np.min(ke, axis=1) for ke in k_euclid]
    wcss = [sum(d**2) for d in dist]
    tss = sum(pdist(dataframe.values)**2)/dataframe.values.shape[0]
    bss = tss - wcss
    plt.plot(bss)
    plt.show()
 
elbow(z_score_df,15)
```
